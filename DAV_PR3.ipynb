# Practical-3 | Apply fltering, sorting, aggregation, and GroupBy operations on real-world datasets.

Step 1 : Import the pandas Libary

```python
import pandas as pd
```

**Step 2: Create a smaple data**

```python
data ={
    "Emp_Id":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

**Step 3: Create a Dataframe**

```python
df = pd.DataFrame(data)
df
```

**Step 4: Display the Dataset**

```python
# Display the complete DataFrame
df
```

**Step 5: Filter Data**

```python
# Select only employees working in the IT department.
#The conditon retruns rows where Department is equal to "IT".

df[df['Department']=="IT"]
```

 Task 1: Display Employee with Salary greater than 50000.

```python
# df["Salary"] → selects the Salary column
# df → your DataFrame (employee data)

df[df["Salary"]>50000]
```

Task 2: Display Employees from surat.

```python
# df → your DataFrame
# Display employees who live in Surat

df[df["City"]=="Surat"]
```

Task 3: Apply Mutliple conditions

```python
# Display employees who are from the IT department
# AND whose salary is greater than 50,000

df[(df["Department"]=="IT")&(df["Salary"]>50000)]

# Both conditions must be True:
# 1. Department must be "IT"
# 2. Salary must be greater than 50,000
```

Step 6: Sort data

Task 1: Sort Salary in Ascending Order.

```python
# Sort the employee data by Salary
# By default, the salary is sorted in ascending order
# (lowest salary to highest salary)

df.sort_values(by="Salary")
```

Task 2: Sort Salary in descending Order.

```python
# Sort the employee data based on the Salary column
# ascending=False means sort from highest salary to lowest salary
# So, the employee with the highest salary will appear first

df.sort_values(by="Salary",ascending=False)
```

Task 3: Sort by Department and Salary.

```python
# Sort the employee data using two columns:
# 1. Department → sort in ascending order (A to Z)
# 2. Salary → sort in descending order (highest to lowest)

df.sort_values(by=["Department","Salary"], ascending=[True,False])
```

Step 7: Perform Aggregation

Task 1: Calculate Total Salary.

```python
# Calculate the total salary of all employees
# df["Salary"] → selects the Salary column
# .sum() → adds all the salary values together

df["Salary"].sum()
```

Task 2: Calculate Average Salary.

```python
# Calculate the average salary of all employees
# .mean() → calculates the average (mean) of the salary values

df["Salary"].mean()
```

Task 3: Find Highest Salary.

```python
# Find the highest salary among all employees
# .max() → finds the largest value in the Salary column

df["Salary"].max()
```

Task 4: Find the lowest Salary.

```python
# Find the lowest salary among all employees
# .min() → finds the smallest value in the Salary column

df["Salary"].min()
```

Task 5: Count Employees

```python
# Count the total number of employee IDs
# .count() → counts the number of non-empty values in the column

df["Emp_Id"].count()
```

Step 8: Perform GroupB operations

Task 1: Average Salary Department-wise

```python
# Group employees according to their Department
# df.groupby("Department") → creates separate groups for each department
# ["Salary"] → selects the Salary column from each department
# .mean() → calculates the average salary for each department

df.groupby("Department")["Salary"].mean()
```

Task 2: Minimum Salary Department-wise

```python
# Group employees by Department
# ["Salary"] → selects the Salary column for each department
# .min() → finds the lowest salary in each department

df.groupby("Department")["Salary"].min()
```

Task 3: Count Employees by Department-wise

```python
# Group employees by Department
# ["Emp_Id"] → selects the Employee ID column
# .count() → counts the number of Employee IDs in each department

df.groupby("Department")["Emp_Id"].count()
```

Task 4: Total Salary Department-wise

```python
# Group employees according to their Department
# ["Salary"] → selects the Salary column
# .sum() → adds all salaries within each department

df.groupby("Department")["Salary"].sum()
```

Task 5: Maximum Salary Department-wise

```python
# Group employees according to their Department
# ["Salary"] → selects the Salary column for each department
# .max() → finds the highest salary in each department

df.groupby("Department")["Salary"].max()
```

Task 6: Group By City Average Salary

```python
# Group employees according to their City
# ["Salary"] → selects the Salary column for each city
# .mean() → calculates the average salary for each city

df.groupby("City")["Salary"].mean()
```

Step 9: Multiple Aggregation Functions

```python
# Group employees by Department
# ["Salary"] → selects the Salary column for each department
# .agg() → applies multiple functions to the Salary column

df.groupby("Department")["Salary"].agg(["count","sum","mean","min","max"])

# count → counts the number of salary values (employees)
# sum   → calculates the total salary
# mean  → calculates the average salary
# min   → finds the lowest salary
# max   → finds the highest salary
```

Step 10: Save Result

```python
# Save the DataFrame as a CSV file
# "Employee_Data.csv" → name of the CSV file to be created
# index=False → does not save the DataFrame index as an extra column

df.to_csv("Employee_Data.csv",index=False)
print("Dataset create successfully.")
```
