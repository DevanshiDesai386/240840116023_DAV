# Practical-4 Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in dataset.

Step1: Import Required Libraries

```python
#Import the Pandas Library
import pandas as pd
#Import the Metplotlib library for creating graphs and charts.
import matplotlib.pyplot as plt
```

Step2:Create the Dataset.

```python
data = {
    'Months': ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],
    'Units_Sold':[120,135,150,145,160,175,180,170,190,210,230,250],
    'Revenue': [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Profit':[20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
    }
```

Step3:Create DataFrame

```python
df = pd.DataFrame(data)
df
```

Step3:Inspect the Dataset

```python
#Display the first five records.
df.head()
```

```python
#Display the number of rows and columns.
df.shape
```

```python
#Display all the columns in the dataset.
df.columns
```

```python
#Display information about the dataset.
df.info()
```

Step4:Generate Descriptive Statistics

```python
#Generate descriptive statistics of all columns.
df.describe()
```

Step5:Calculate Mean

```python
#Calculate the averge number of units sold per month.
average_units = df['Units_Sold'].mean()
#Display the result.
print("Average Units Sold:", average_units)
```

Step6:Calculate Median

```python
#Calculate the middle value of the Units_Sold column.
median_units = df['Units_Sold'].median()
#Display the result.
print("Median Units Sold:", median_units)
```

Step7:Calculate Maximumand minimum

```python
#Find the higest number of units sold in a months.
maximum_units = df['Units_Sold'].max()
#Find the lowest number of units sold in a months.
minimum_units = df['Units_Sold'].min()
#Display the result.
print("Maximum Units Sold:", maximum_units)
print("Minimum Units Sold:", minimum_units)
```

Step8:Calculate Standard Deviation

```python
#Calculate the standard deviation od units sold.
std_units = df["Units_Sold"].std()
#Display the result.
print("Standard Deviation of Units Sold:", std_units)
```

Step9:Find the Month with Highest Sales

```python
#Find the index of the row having the maximum revenue.
highest_sales_index = df["Revenue"].idxmax()

#Retrive the month corresponding to that index.
highest_sales_month = df.loc[highest_sales_index, "Months"]

#Retrive the maximun revenue.
highest_revenue = df.loc[highest_sales_index, "Revenue"]

#Display the result.
print("Highest Sales Month:", highest_sales_month)
print("Highest Revenue:", highest_revenue)
```

Step10:Find the Month with the Lowest Sales

```python
#Find the index of the row having the minimum revenue.
lowest_sales_index = df["Revenue"].idxmin()

#Retrive the month corresponding to that index.
lowest_sales_month = df.loc[lowest_sales_index, "Months"]

#Retrive the minimum revenue.
lowest_revenue = df.loc[lowest_sales_index, "Revenue"]

#Display the result.
print("Lowest Sales Month:", lowest_sales_month)
print("Lowest Revenue:", lowest_revenue)
```

Step11:Calculate Total Revenue

```python
#Calculate the total revenue generated guring the year.
total_revenue = df["Revenue"].sum()
#Display the result.
print("Total Revenue Generated:", total_revenue)
```

Step12:Calculate Total Profit

```python
total_profit = df["Profit"].sum()
#Display the total profit.
print("Total Profit:", total_profit)
```

Step13:Identify the Sales Trend

```python
#Create a line chart to visualize monthly revenue.
plt.figure(figsize=(10, 5))

#Plot month on the x-axis and revenue on the y-axis.
plt.plot(df["Months"], df["Revenue"], marker='o', linestyle="--")

#Add a title to the graph.
plt.title("Monthly Revenue Trend")

#Display Month on x-axis.
plt.xlabel("Months")

#Display Revenue on y-axis.
plt.ylabel("Revenue")

plt.xticks(rotation=45)

#Display the graph.
plt.show()
```

Step14:Analyze Profit Trend

```python
#Create a line chart to visualize monthly Profit.
plt.figure(figsize=(10, 5))

#Plot month on the x-axis and revenue on the y-axis.
plt.plot(df["Months"], df["Profit"], marker='o', linestyle="--")

#Add a title to the graph.
plt.title("Monthly Profit Trend")

#Display Month on x-axis.
plt.xlabel("Months")

#Display Profit on y-axis.
plt.ylabel("Profit")

plt.xticks(rotation=45)

#Display the graph.
plt.show()
```

Step15:Compare Revenue and Profit

```python
#Create a line chart to compare revenue and profit.
plt.figure(figsize=(10, 5))

#Plot Revenue.
plt.plot(df["Months"], df["Revenue"], marker='o', label="Revenue")

#Plot Profit
plt.plot(df["Months"], df["Profit"], marker='o', label="Profit")

#Add a title to the graph.
plt.title("Comparison of Revenue and Profit")

#Display Month on x-axis.
plt.xlabel("Months")

#Display Amount on y-axis.
plt.ylabel("Amount")

#Display the legend to indetify tow line
plt.legend()

plt.xticks(rotation=45)

#Display the graph.
plt.show()
```

Step16:Identify Growth in Sales

```python
#Calculate the percentage change in revenue
#from one month to next month.
df["Revenue_Growth_%"] = df["Revenue"].pct_change() *100
#Display the update dataset.
df
```

Step17:Identify Months with increasing Revenue

```python
#Fliter the dataset to display only those months
#where revenue growth is positive.
increasing_sales = df[df["Revenue_Growth_%"]>0]
#Display the result.
print("Months with Increasing Revenue:")
print(increasing_sales["Months"])
```
