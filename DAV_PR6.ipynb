# **Practical-6**  Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).

Step1 Import require libraries numpy, matplotlib.

```python
# Import Pandas for data manipulation and analysis
import pandas as pd
# Import NumPy for numerical calculations
import numpy as np
# Import Matplotlib for creating graphs and visualizations
import matplotlib.pyplot as plt
# Import Seaborn for statistical data visualization
import seaborn as sns
```

Step2 Create a dataset.

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Product': ['Laptop','Mobile' , 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Sales': [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Units_Sold': [12,30,25,14,35,30,18,38,32,21,45,40]
}
# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)
# Display the DataFrame
df
```

Task2- Create a Line Chart

How did the company's sales change from January to December?

```python
# Create a line chart
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['Sales'], marker='o', linestyle='-', color='b')
plt.title('Company Sales Over Time')
plt.xlabel('Month')
plt.ylabel('Sales')
```

**Interpretation:**

The line chart shows that the company's sales increased gradually from January to December. Sales were lowest in January at ₹1,20,000 and highest in December at ₹2,50,000. Overall, the graph indicates a positive growth trend in the company's sales throughout the year.

Task3- create a Bar Chart

Which product generated how much total sales?

```python
#Create a bar chart
plt.figure(figsize=(6, 4))
plt.bar(df['Product'], df['Sales'], color='g')
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()
```

**Interpretation:**

The bar chart compares the sales of different products. The sales values vary for Laptop, Mobile, and Tablet across different months. Overall, the chart helps in comparing the sales performance of the three products.

Task4- Create a Histogram

How are the indivudual sales values distributed?

```python
# Create a histogram
plt.figure(figsize=(6, 4))
# Use a single color from the pastel palette
plt.hist(df['Sales'], bins=6, color=sns.color_palette('pastel')[0], edgecolor='black')
plt.title('Distribution of Sales Values')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

**Interpretation:**

The histogram shows the distribution of individual sales values. It indicates that the sales values are spread across different ranges, with higher sales values occurring more frequently toward the later part of the year. This shows an overall increase in sales.

Task5 - Create a Scatter Plot

Is there a relationship between the number of units sold and total sales?

```python
# Create a scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(df['Units_Sold'], df['Sales'], color='blue', marker='*')
plt.title('Relationship between Units Sold and Sales')
plt.xlabel('Units Sold')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

**Interpretation:**

The scatter plot shows the relationship between units sold and total sales. The points generally move upward as the number of units sold increases. This indicates a positive relationship, meaning that higher units sold are generally associated with higher sales.

Task6 - Create a Pie Chart

What percentage of total sales comes from each region?

```python
# Create a pie chart
region_sales = df.groupby('Region')['Sales'].sum()
labels = region_sales.index
sizes = region_sales
plt.figure(figsize=(4, 4))
plt.pie(sizes, labels=region_sales.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Percentage of Total Sales by Region')
plt.axis('equal')
plt.show()
```

**Interpretation:**

The pie chart represents the percentage contribution of each region to the company's total sales. It shows that the sales contribution differs among the North, South, East, and West regions. The region with the largest section of the pie contributes the highest share of total sales.
